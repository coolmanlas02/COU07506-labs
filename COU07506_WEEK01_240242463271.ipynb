```python
import matplotlib.pyplot as plt
import numpy as np

n = np.arange(-5, 11)
delta = (n == 0).astype(float)  # unit impulse
step = (n >= 0).astype(float)  # unit step
ramp = n * step  # unit ramp
expo = 0.8**n * step  # real exponential

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))
for ax, x, title in zip(
    axes, [delta, step, ramp, expo], ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]
):
  ax.stem(n, x, basefmt="k-")
  ax.set_title(title)
  ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

```python
n = np.arange(-5, 11, dtype=float)
step = (n >= 0).astype(float)
a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, a in zip(axes, a_values):
  x = (a**n) * step
  ax.stem(n, x, basefmt="k-")
  ax.set_title(f"({a})^n u(n)")
  ax.set_xlabel("n")
  ax.set_ylabel("Amplitude")
  ax.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# TASK 1.2 - PART 2: COSINE SIGNALS

import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 41)


x1 = np.cos(2 * np.pi * n / 8)
x2 = np.cos(2 * np.pi * n / 7.5)


plt.figure(figsize=(10, 4))

plt.stem(n, x1, basefmt="k-", label="cos(2πn/8)")
plt.plot(n, x2, "ro--", label="cos(2πn/7.5)")

plt.title("Comparison of Two Cosine Signals")
plt.xlabel("n")
plt.ylabel("Amplitude")

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

2. Why a Stem Plot is Appropriate for x(n)

A discrete time signal x(n) is defined only at integer values of the sample index n it has no defined values between adjacent samples. A stem plot accurately reflects this by drawing isolated markers at each integer n, whereas a tinuous line plot interpolates between points and creates the false impression that the signa. ists continuously in time.
